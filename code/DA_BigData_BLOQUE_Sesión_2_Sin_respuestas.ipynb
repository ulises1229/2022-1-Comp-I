### Añadir Colores y modificar el tamaño del gráfico

````
#importar librería de numpy
import numpy as np

# Almacenar variables x = años servicio y y = salario
x = df.service
y = df.salary
lab = ['Male', 'Female']

plt.figure(figsize=(10, 5))


# Arreglo de colores
colores = np.arange(len(x))

col = df['rank']
cols = []

for i in range(len(col)):
    if col[i] == 'Prof':
        cols.append(0)
    elif col[i] == 'AsstProf':
        cols.append(1)
    elif col[i] == 'AssocProf':
        cols.append(2)

print(cols)

# Títulos: gráfico y ejes
plt.title("Gráfico de Dispersión Años de servicio vs Salario")
plt.xlabel('Años de servicio')
plt.ylabel('Salario Anual')

# Agregar un scatter con colores y transparencia
plt.scatter(x, y, c=cols, alpha = 0.7, s = (x*10), label = "Profesor")
plt.legend(loc='best')

#plt.show()
plt.savefig("dispersión2.png", dpi=600, bbox_inches='tight')
````

## 4. Box Plots
Los gráficos de cajas, también conocidos como boxplots, se utilizan para mostrar la distribución de los datos a través de sus cuartiles.

````
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.figure(figsize=(10, 5))
plt.boxplot(data, vert=True, patch_artist=True)
plt.title('Gráfico de Cajas')
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.grid(True)
plt.show()
````

````
import matplotlib.pyplot as plt
import pandas as pd

# Leer el dataframe mpg.csv
mpg = pd.read_csv("https://raw.githubusercontent.com/ulises1229/INTRO-PYTHON-ENESJ/master/data/mpg.csv")

# Preprocesamiento de datos
data_by_class = [mpg[mpg['class'] == cls]['hwy'] for cls in mpg['class'].unique()]
classes = mpg['class'].unique()

# Mejorar el boxplot para todas las clases de automóviles
plt.figure(figsize=(14, 8))
plt.boxplot(data_by_class, notch=True, patch_artist=True,
            boxprops=dict(facecolor='lightblue', color='blue'),
            whiskerprops=dict(color='blue'),
            capprops=dict(color='blue'),
            medianprops=dict(color='red'),
            flierprops=dict(marker='o', color='red', alpha=0.5))

# Agregar títulos y etiquetas
plt.title("Consumo en autopista por clase de automóvil", fontsize=14)
plt.xlabel("Clase de automóvil", fontsize=12)
plt.ylabel("Rendimiento en autopista (mpg)", fontsize=12)
plt.xticks(ticks=range(1, len(classes) + 1), labels=classes, rotation=45)

# Agregar un fondo gris claro
plt.gca().set_facecolor('whitesmoke')

# Añadir líneas de cuadrícula
plt.grid(True, linestyle='--', alpha=0.7)

# Exportar imagen
plt.savefig("mpg_clases_mejorado.tiff", dpi=300, bbox_inches='tight')

# Mostrar el gráfico
plt.show()
````

## Gráficos de Área

Los gráficos de área se utilizan para mostrar la evolución de una variable a lo largo del tiempo.

````
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 5))
plt.fill_between(x, y1, alpha=0.5, label='Seno')
plt.fill_between(x, y2, alpha=0.5, label='Coseno')
plt.title('Gráfico de Área')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend()
plt.grid(True)
plt.show()
````

# 2. Análisis Exploratorio de Datos

En esta sección se realiza un análisis exploratorio de datos utilizando pandas, Matplotlib y Seaborn.

## Importar bibliotecas y cargar datos

````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Crear un DataFrame de ejemplo
np.random.seed(42)
data = pd.DataFrame({
    'A': np.random.randn(100),
    'B': np.random.randn(100) * 50 + 100,
    'C': np.random.randn(100) * 5 + 50,
    'D': np.random.randint(1, 100, 100)
})

data.head()
````

## Descripción de los datos

````
data.describe()
````

## Visualización de distribuciones

````
plt.figure(figsize=(10, 5))
sns.histplot(data['A'], kde=True)
plt.title('Distribución de A')
plt.show()
````

````
plt.figure(figsize=(10, 5))
sns.histplot(data['B'], kde=True)
plt.title('Distribución de B')
plt.show()
````

## Visualización de relaciones

````
plt.figure(figsize=(10, 5))
sns.scatterplot(x='A', y='B', data=data)
plt.title('Relación entre A y B')
plt.show()
````

## Matriz de correlación y heatmap

````
corr = data.corr()
plt.figure(figsize=(10, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Matriz de Correlación')
plt.show()
````

## Visualización de distribuciones conjuntas

````
plt.figure(figsize=(10, 5))
sns.jointplot(x='A', y='B', data=data, kind='hex', cmap='Blues')
plt.show()
````

## Visualización de distribuciones múltiples

````
sns.pairplot(data)
plt.show()
````

# Análisis Exploratorio de Datos con un set de Datos Real


## Importar bibliotecas y cargar datos

````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Leer el dataframe mpg.csv
mpg = pd.read_csv('https://raw.githubusercontent.com/ulises1229/INTRO-PYTHON-ENESJ/master/data/mpg.csv')

mpg.head()
````

## Descripción de los datos

````
mpg.describe()
````

## Visualización de distribuciones

````
plt.figure(figsize=(14, 7))
plt.subplot(1, 2, 1)
sns.histplot(mpg['hwy'], kde=True)
plt.title('Distribución de Consumo en Autopista (hwy)')
plt.xlabel('Consumo en Autopista (mpg)')
plt.ylabel('Frecuencia')

plt.subplot(1, 2, 2)
sns.histplot(mpg['cty'], kde=True)
plt.title('Distribución de Consumo en Ciudad (cty)')
plt.xlabel('Consumo en Ciudad (mpg)')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
````

## Visualización de relaciones

````
plt.figure(figsize=(10, 6))
sns.scatterplot(x='displ', y='hwy', data=mpg, hue='class')
plt.title('Relación entre Desplazamiento y Consumo en Autopista')
plt.xlabel('Desplazamiento del Motor (litros)')
plt.ylabel('Consumo en Autopista (mpg)')
plt.legend(title='Clase de Automóvil')
plt.show()
````

## Boxplot de Consumo en Autopista por Clases de Automóviles

````
plt.figure(figsize=(14, 8))
sns.boxplot(x='class', y='hwy', data=mpg, palette='Set3')
plt.title('Consumo en Autopista por Clase de Automóvil')
plt.xlabel('Clase de Automóvil')
plt.ylabel('Consumo en Autopista (mpg)')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
````

## Matriz de correlación y heatmap

image.png

````
corr = mpg[['displ', 'hwy', 'cty', 'cyl', 'year']].corr()
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Matriz de Correlación')
plt.show()
````

## Visualización de distribuciones conjuntas

````
plt.figure(figsize=(10, 6))
sns.jointplot(x='hwy', y='cty', data=mpg, kind='hex', cmap='Blues')
plt.show()
````

## Visualización de distribuciones múltiples

````
sns.pairplot(mpg)
plt.show()
````
